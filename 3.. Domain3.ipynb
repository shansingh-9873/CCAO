# 2.. Choosing Between Chat, Projects, Research, and Artifacts

```python
'''


Hi there!
In this lecture, we are going to learn how to choose the right workspace in Claude before starting your work.

You already know what chats, Projects, Research, and Artifacts are.
But the important question is: **When should you use each one?**



    1 ) Chat – How Long Do You Need the Context?
        Always Start by asking yourself one question:
            **Will I need this information only once or repeatedly?**

            & simple hack is : **Use a Chat for one-time tasks.**

            For example :

            - Asking a quick question.
            - Drafting a short email.
            - Getting an explanation of a concept.

        If you don't need the same background information in future conversations, a normal chat is enough.


    2 ) Project 
        **Use a Project for ongoing work.**

        Imagine you are managing a six-week client project.
        So In CLAUDE , You may repeatedly use the same company background, reference documents, and writing instructions.

        Instead of providing this information in every conversation, add it to a Claude Project.
        Then You can then start multiple chats within that Project and reuse the shared context.

        & the big benefit is that : Projects can also be shared with colleagues but on Team and Enterprise plans...





    3 ) When Should You Use Research?

        Research is different from Chat and Projects.
        It is not another workspace , It is a way to investigate a topic using multiple connected searches.

        Claude searches for information, explores relevant sources, and prepares a report with citations that you can verify.
        For example:

        - Comparing three competitors.
        - Researching a new government regulation.
        - Understanding the latest market trends.


        [show in CLAUDE AI]
        Research can also use connected tools, such as Gmail and Google Docs, where supported..
        
        
        Q. Now, when should you use it?

        Use Research when your question requires substantial investigation.
        For example, if you could answer something with one simple search, a normal chat is usually enough.

        But if the research would take you 30 minutes or more to complete manually, Research may be worth using.
        Remember, Research can take longer and consume your usage limits faster..
        
        
        

    4 ) When Should You Use Artifacts?
        Use an Artifact when you are creating content that you want to edit, improve, or reuse over multiple rounds.

        For example:

            - A business report.
            - A presentation.
            - A dashboard.
            - A document or application.

    Instead of asking Claude to regenerate the entire content every time, you can request specific changes to the existing Artifact.
    This makes ongoing editing easier.





    Real-World Example : 
    For example, imagine you are preparing a market research report for a client.

        - **Project:** Stores the client's background and reference documents.
        - **Chat:** Helps you discuss specific tasks.
        - **Research:** Collects information about competitors and market trends.
        - **Artifact:** Holds the final report that you can edit and share.




    Let's quickly summarize.

         •  Chat :     For quick, one-time conversations.
         •  Project :  For ongoing work where you repeatedly need the same context.
         •  Research:  For questions that require deeper investigation and multiple searches.
         •  Artifact:  For content that you want to edit and reuse.



        Remember these two questions before starting:

        1. How long will I need this context?
        2. What do I need Claude to do?




'''
```

# 3.. Meet Claude Models : OPUS , SONNET , HAIKU , FABLE

```python
'''


[ show via CLAUDE ]

Hi there !
Every conversation in Claude runs on a model, and you can select that model using the model picker..
But which model should you choose ?

So In this lecture, we are going to understand how to choose the right Claude model for your work..
Whether
    •  Opus
    •  Sonnet
    •  Haiku
    
        & their recent & most recent model , ie FABLE..
        But exam guide usually covers these 3 :
        •  Opus
        •  Sonnet
        •  Haiku


To understand that, let's first look at three things your model choice affects:
See Claude offers different models, and each one has a different balance of these three factors.

    •  Speed -  How quickly Claude responds.
    •  Usage -  How much of your usage limits the model consumes.
    •  Capability -  How well it handles complex tasks.




    One thing to remember : Model names and their versions can change over time, 
                            and the models available in your model picker depend on your subscription plan.



[come to PPT]

Let's understand each model..

    1 ) Haiku – Fast and Lightweight
        Haiku is designed for speed and relatively simple tasks.
        It is generally the fastest and most economical option.
        For example, imagine you receive hundreds of customer emails and want Claude to classify them into categories 
                     such as complaints, enquiries, and feedback...
                     
        Haiku is a suitable choice for this type of repetitive task.




    2 ) Sonnet – Balanced Performance
        Sonnet provides a balance between speed and capability.
        It is suitable for everyday tasks such as:

            - Writing and editing content.
            - Analyzing data.
            - Summarizing documents.
            - General coding tasks.

        So For most regular work, Sonnet is a good starting point.






    3 ) Opus – Complex Tasks

        Opus is designed for more demanding tasks that require deeper reasoning and handling multiple steps.
        For example, imagine you want Claude :
                    To analyze a large business problem, evaluate different strategies, and prepare a detailed recommendation.

        Opus may be a better choice for this type of complex work.




    4 ) FABLE - it is Beyond Opus?
        Anthropic also has a higher capability class called Mythos.
        Fable belongs to this Mythos class and is designed for highly advanced tasks.

        However, access to these models is more restricted and depends on the subscription or access arrangements available.






    Q. How to Choose the Right Model ?
    Ans : Now, Let's bring everything together..

            Imagine you have 3 different tasks :
                - Classifying hundreds of customer emails: Choose Haiku..
                - Writing a business report or analyzing sales data: Choose Sonnet..
                - Solving a complex business problem involving multiple steps: Choose Opus..




    Notice that we don't always need the most powerful model..
    Using a more capable model for a simple task may consume more of your usage limits without providing much additional benefit.

    One more impotant thing : more capable models generally require more resources and may take longer to respond.



    Let's quickly summarize.

    1. **Haiku:** Fast and economical, suitable for simple and repetitive tasks.
    2. **Sonnet:** Balanced speed and capability, suitable for everyday work.
    3. **Opus:** Designed for complex tasks requiring deeper reasoning.
    4. **Mythos:** A higher capability model class that includes Fable.
    5. **Model selection:** Choose based on task complexity, speed, and available usage limits.

    Remember, **the most powerful model is not always the best choice. 
    Choose the model that matches the complexity of your task.**


'''
```

# 4.. When to use which Models !

```python
'''


Hi there!
In the previous lecture, we learned about three Claude model families: Haiku, Sonnet, and Opus.
But how do you decide which model to use for a particular task?

Let's understand this with a simple example.

    Imagine you have a quick task, but you select the most powerful model available 
    because you think it will give you the best result.

    The problem is that you may end up spending more time and usage limits than necessary.

    Instead, ask yourself three simple questions.

     1 ) How Much Thinking Does the Task Require?
        First, understand whether your task requires simple processing or deeper reasoning.
        For example:

        - Reformatting a list.
        - Drafting a routine email.
        - Extracting information from documents.

        These are relatively simple tasks, so a lighter model like Haiku may be sufficient.

        But imagine you need to analyze why your company's sales have declined or compare three different business strategies.
        These tasks require deeper reasoning, so a more capable model like Sonnet or Opus may be a better choice.




    2 ) Does Speed or Volume Matter?
        Now, consider how frequently you need to perform the task.

        For example, generating one detailed business report is different from classifying 400 customer emails.
        When you have hundreds of repetitive tasks, a faster and more economical model can save considerable time and usage.
        So, for simple, repetitive, high-volume work, consider starting with Haiku.




     3 ) What Happens If the Answer Is Wrong?
        This is another important question.

        Imagine Claude prepares an internal email draft.
        If something is incorrect, you can easily fix it before sending.

        But what if Claude prepares a financial recommendation that someone might act upon?
        So The consequences could be much more serious.


        •  So, when the cost of an incorrect answer is high, consider using a more capable model 
            and make sure the output is properly verified.

        •  Always Remember, choosing a powerful model does not replace human review.



     4 ) Understanding Effort Levels
         [show CLAUDE AI Interface]
        Apart from choosing a model, Claude also allows you to adjust its reasoning effort.
        Depending on the model and interface, you may see options such as:

        - Low
        - Medium
        - High
        - Extra High
        - Max



        •  Higher effort generally means Claude spends more resources reasoning through a task.
            This can produce more thorough answers, but it may also increase response time and usage.

        •  For routine tasks, low or medium effort may be sufficient.

        •  For complex analysis, you can consider higher effort levels.






    5 ) A Practical Approach to Model Selection


    •  For simple tasks   => Start with a lighter model.
            Check the output and move to a more capable model only if the result isn't good enough.

    •  For complex tasks  => Start with a more capable model.
            Once you get a satisfactory result, try reducing the effort level or using a lighter model to see 
            whether it can handle the same task.






    Two Common Mistakes to Avoid :

    •  Use the most powerful model for everything :
            This can consume your usage limits unnecessarily and slow down simple tasks.



    •  Using a lightweight model for every task just to save usage.**
            If the task requires complex reasoning, the output may not be good enough, 
            and correcting it could cost you more time.








'''
```

# 5.. Context Window : Limit & how to handle it !

```python
'''



Hi there!

In this lecture, we are going to understand the context window in Claude and how to manage it effectively.
                 See context window is everything Claude can work with at one moment in a conversation.

**Context Window:** The amount of information Claude can work with at one moment.



But why is this important ?
Because as your conversations grow longer and you add more files and instructions, the available context gets used up.



Let's understand what fills the context window and what happens when it gets close to its limit.

    1 ) What Fills the Context Window?
        Several things contribute to the context Claude uses.
        For example:

            - Your messages and questions.
            - Claude's responses.
            - Uploaded documents and files.
            - Project instructions.
            - Connected tools and integrations.

    Even a single large document can consume a significant amount of context.
    Similarly, enabling many tools and connectors can also use additional context.
    That's why it is useful to keep only the files, instructions, and tools that your task actually needs.





     2 ) What Happens When a Chat Gets Too Long?
         Now, imagine you have been working with Claude in the same conversation for several hours.
         You have shared multiple documents, asked many questions, and received lengthy responses.

         Eventually, the conversation may approach its context limit.
         So CLAUDE will compact/summarize earlier messages to make room for new ones.

             This helps Claude continue working without requiring you to manually start a new conversation.
             However, there is one limitation.

             **A summary is not the same as the original conversation.**
               Claude may preserve the important points but shorten earlier details.




    3 ) How Does Retrieval Help in Projects?

        Instead of one conversation , you may have too many documents & u have stored it in a Project.
        This is where Retrieval-Augmented Generation, commonly called RAG, becomes useful.

        Normally, Claude works with information available in its context.
        But With retrieval, Claude can search through your project documents and use the relevant sections 
                            instead of loading every document at once.



        For example, imagine your Project contains 50 business reports.
                     You ask Claude about the sales performance of a particular product.

                     Instead of using all 50 reports, Claude can retrieve the relevant information from the documents 
                     that relate to your question.

                     This helps manage large collections of information.


                     See You can also improve retrieval by:
                        - Giving your files clear names.
                        - Keeping related documents together.
                        - Mentioning the specific document in your question when necessary.






    4 ) Context Window vs Usage Limits
        There is one important distinction I want you to understand.
        These two terms are not the same.

        •  Context Window :  The amount of information Claude can work with at one moment.

        •  Usage Limits : The amount of Claude usage available to you over a particular period.

            & one more thing : When you start a new chat  , it gives you a fresh conversation context..
                               But it does not automatically reset your usage limits.
      






    5 ) Restart, Summarize, or Persist?
        Now, what should you do when a conversation becomes too long?

        You have three practical options.

        I ) Restart – Start a new chat.**
            Use this when you are switching to a completely different topic 
                        or 
            the existing conversation has gone off track.


            For example, if you were discussing a marketing strategy and now want to work on Python programming, 
                         So in such scanrio , starting a fresh chat makes sense.




        II ) Summarize – Carry forward important information.**
             Sometimes, you want to continue the same work but don't need the entire conversation.
             In that case, ask Claude to prepare a summary containing:

                - Important decisions.
                - Key instructions.
                - Relevant facts.
                - Pending tasks.

            Copy that summary into a new chat and continue working.




        III ) Persist – Save information for future use.**
              If certain information will be useful repeatedly, 
              store it in a Project's knowledge or Claude's available memory features.

            For example, you can save your company background, writing guidelines, or frequently used instructions.
                This avoids providing the same information repeatedly.








Let's quickly summarize.

•  Context Window:** Messages, responses, files, instructions, and tools all consume context.
•  Long Conversations:** Automatic context management can summarize earlier messages, but some details may be lost.
•  Retrieval:** Helps Claude find relevant information from large collections of project documents.
•  Usage Limits:** Starting a new chat does not reset your usage limits.
•  Three Choices:** Restart for a fresh topic, summarize to carry important information forward, or persist information you will reuse.

Remember, **don't keep adding information to a conversation without considering what Claude actually needs.**

Choose the right approach to keep your conversations organized and your context useful.









'''
```
