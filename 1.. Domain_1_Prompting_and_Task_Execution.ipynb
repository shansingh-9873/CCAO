# 2.. Anatomy of a Strong Prompt


````python
'''

Hi there 👋,

In this lecture, we are going to understand the anatomy of a strong prompt.




Here is a structure for a strong prompt :
    Role + Task + Context + Constraints + Format + Examples.

Let's understand each part one by one.

1 ) ROLE 
    First, let's talk about the role.
    When you assign a role to CLAUDE , it tells Claude who it should act as. 

    For example :
        instead of simply saying, "Plan a trip to DUBAI."" 
        you can provide a role and say :
        "You are an experienced travel planner who specializes in budget-friendly family trips. Plan a trip to DUBAI."
        
        Assigning a role helps Claude focus on the expected tone and behaviour for the task.
        Now Claude has a clearer idea of the perspective it should use.



2 ) TASK 
    Task is the instruction itself..
    The task tells Claude exactly what you want it to do. 
    Hint : For task , Always use precise verbs for it 
    eg , Do you want Claude to summarize, compare, classify, draft, analyze, or organize something? 

    • In weak prompt , It asks Claude to help with something..
    • But In strong prompt , task tells Claude the exact task with one clear job 
    
    Instead of saying: ❌
    "Help me with my Jaipur trip."

    You can give Claude a specific instruction: ✅
    "Create a 3-day travel itinerary covering popular tourist attractions, shopping markets, and local food spots." 


    See The first instruction is quite broad bcz here Claude has to figure out what kind of help you need.
    But the second instruction clearly defines the job.

    So, whenever you write a prompt, make sure you tell Claude exactly what you want it to accomplish.
    




3 ) CONTEXT
    The third part is context. 
    Context explains the background that CLAUDE doesn't know..
    CONTEXT Tells Claude :
        • what the situation is, 
        • who the audience is, 
        • what the audience already knows. 
        
    CONTEXT matters otherwise CLAUDE may have to guess the background or make assumptions about the 
    purpose of the task.



    Lets understand how u can provide context :

    "I am planning a 3-day DUBAI trip with my family. 
    We are four people, including two children. 
    We will be travelling from Delhi by FLIGHT , and this is our first visit to DUBAI."

    Now Claude understands your situation.

    It knows :
        • how many people are travelling, 
        • whether children are involved, 
        • where you are travelling from, 
        • and whether you have visited DUBAI before.






4 ) CONSTRAINTS
    The fourth part is constraints. 
    Constraints define the boundaries/rules that Claude should follow..
    
    For example, you can specify 
        • The maximum budget
        • the level of detail, 
        • What should or should not be included. 
        
        Let's understand this using our DUBAI trip example..
        
        You can tell Claude :

        "Keep the total trip budget under ₹20,000. Avoid expensive restaurants and luxury hotels. 
        Include no more than four tourist attractions per day, and prefer family-friendly places."
        Now Claude understands the limitations it needs to respect while preparing the itinerary.
        


        • You can also explain the reason behind an important constraint. 
            For example, instead of only saying, "Include no more than four tourist attractions per day.,"
            
            you can say this : 
            
            "Include no more than four tourist attractions per day because we are travelling with children 
             and want to avoid a hectic schedule." 
             
             • Without ❌ reason , CLAUDE will guess 
             • Giving the reason ✅ helps Claude understand the goal behind the instruction , 
               it knows the limit & it knows WHY ?
               So CLAUDE works better when it knows the reason bcz reason fills gaph sensibly
               
               ie give reason = better decision




5 ) FORMAT
    The fifth part is format. 
    Format describes the shape/sructure of the answer you want. 
    • Do you want a table with named columns, a short email, a report, or 
    • a set of bullet points? 
    
    
    If you already know the structure you need, state it directly.



    For example, you could ask : 
    Present the itinerary in a table with four columns: Day, Time, Place to Visit, and Estimated Cost."

    This tells Claude exactly how to organize the information.
    The more clearly you describe the expected format, the less Claude has to decide on its own.





6 ) Examples
    ( ie show sample output , dont just explain )
    
    Sometimes, explaining your requirements in words is not enough. 
    Showing Claude an actual example can make your expectations much clearer.

    Examples show Claude what a good answer/result should look like. 
    If you have 2-4 examples of the output you want, include them in the prompt. 
    
    This is especially useful when you need a particular structure , format , tone or a consistent pattern. 
    
    When u attach 2-4 different examples =>> Makes pattern clear for CLAUDE => gives more consistent results


    "Here is an example of how I want the itinerary to look:

    Day 1:

    • Morning: Visit a historical monument.
    • Afternoon: Have lunch at a budget-friendly restaurant.
    • Evening: Explore a local shopping market.

    Follow a similar structure for Day 2 and Day 3."
    Now Claude has a reference to follow while preparing the remaining itinerary.





In the next lecture, we will focus more closely on context and understand why giving Claude the right background 
makes a difference.

'''
````

````python
'''



Scenario: Classifying office emails by priority
    Imagine you receive 50 emails every day..

    Some emails are urgent, some are normal, and some are low priority.

    You want Claude to automatically classify these emails into three categories:
        - High Priority
        - Medium Priority
        - Low Priority
        
        
        
        
1 ) Zero-shot prompting
        Here we give Claude the task without providing any examples.
        
        [SHOW PROMPT] :
        
        
You are an Office Administration Assistant.

Task: Classify the following emails into three categories: High Priority, Medium Priority, and Low Priority.

Context: These emails are received by employees in a corporate office. 
         High-priority emails require immediate attention, 
         medium-priority emails require attention within a reasonable time, 
         and low-priority emails are routine communications.

Constraints:
        - Assign only one priority to each email.
        - Do not ignore important deadlines.
        - Do not change the original meaning of the email.

Format: Present the result in a table with three columns: Email, Priority, and Reason.

Email to classify:
        The client has reported a major issue with the payment system. Please resolve it before 2 PM today.




         We have given Claude the task and instructions without attaching any examples of how to classify emails.
         That is Zero-shot prompting..







2 ) Few-shot prompting
        Now, let's use the same task, but this time provide Claude with a few examples..
        So here u can tell CLAUDE what the High Priority, Medium Priority, and Low Priority means , 
        & for that in general , u have to define the rules.. but rules leave gaps & then again CLAUDE has to guess !
        
        Instead of defining rules , 
        attach 2-4 examples (EMAILS with LABELS) + Attach 1 Edge case example/Ticky example
        
            Make sure , these examples (EMAILS with LABELS) will be real I/Ps , real length , real phrasing
            Edge case : 1 Email that is of medium priority but it is high priority , 
            so this will help CLAUDE to understand where BOUNDARY sits..
            
            now CLAUDE will generalizes it correctly & see what actual pattern is.
            This is Few-shot prompting..
        
        
        
     [SHOW PROMPT]   
        
    You are an Office Administration Assistant.

    Task: Classify emails as High, Medium, or Low Priority.

    Context: These are everyday emails received by employees in a corporate office.

    Examples:

    Example 1:
    Email: "The payment gateway is down. Customers cannot complete transactions. Please investigate immediately."
    Priority: High

    Example 2:
    Email: "Please share the monthly sales report by tomorrow's team meeting."
    Priority: Medium

    Example 3:
    Email: "The office will remain closed this Friday due to maintenance."
    Priority: Low

    Example 4 (Edge Case):
    Email: "Please approve the pending vendor invoice by 5 PM today to avoid a late payment penalty."
    Priority: High



    Now classify this email:
    Email: "Please review the revised project proposal before tomorrow's client meeting."

    Constraints:
        - Assign exactly one priority.
        - Follow the classification patterns in the examples.
        - Do not assume urgency without supporting information.

    Output Format:
    Present the result in a table with three columns:
    Email | Priority | Reason












    WHEN TO USE WHAT :
    
    •  When task is easy to explain or you can describe the JOB in a sentence , then ZERO-SHOT  is fine !
    •  When task have specific pattern / tricky edge cases , that are hard to put into words/rules , use FEW-SHOT...
    
    
    
    
    Where FEW-SHOT WORKS :
    
    Businesses which benefit most from this : 
    
     • Tagging
            When you are sorting something into buckets , ie
            EMAILS into priorities
            TICKETS into Teams
            FEEDBACK into category
        
        
        
     •  Extraction
            When u want to extract something , like :
            When u want names , dates , amt out of document :
                Use FEW-SHOT Prompting , where u can attach one filled-in example..
        
        
        
     •  Repeatable Drafting
            If u write same kind of thing often , ie
                a status often
                a customer reply
            






'''
````

# 5.. Context in Claude & why it Matters !


````python
'''

Hi there 👋,

In this lecture, we are going to focus on little bit more on context in Claude , bcz 
CONTEXT is the key thing when u are working with AI Models !



Q. What is CONTEXT ?
ANS : Context is the background information that helps Claude understand the situation behind your request. 
      It tells Claude 
        • Who the reader is 
        • what the Goal is , 
        • what the audience already knows.
        • Any subject-matter detail
        
        
        
        
Let's understand why this matters.

Imagine that you ask Claude, "Write a short announcement."

    • The instruction does not explain what the announcement is about, 
    • who will read it, or 
    • what the readers already know. 
    
    Claude has to make decisions about those missing details. 
    As a result, the response may not match the situation you had in mind.



    Now compare that with a prompt that explains the background.

    For example, you could say that you are preparing an announcement for customers about a change to a service policy. 
    You can also explain the audience and the purpose of the announcement. 
    With that context, Claude has a clearer picture of what it is being asked to write.


    There are three useful questions to consider when providing context.

    • First: what is the background?
        Explain the situation Claude needs to understand. Give the relevant information about the subject or the task.

    • Second: who is the audience?
        Tell Claude who will read or use the response. 
        For example, the audience might be customers, managers, or people who are already familiar with the subject.

    • Third: what does the audience already know?
        This helps Claude choose an appropriate level of explanation. 
        A response for readers who already understand a subject may be different from a response for 
        people who are encountering it for the first time.


    Context also includes the reason behind the request. 
    Claude can make a more appropriate choice when it understands why you need the output.



For example, compare these two instructions :
    "Keep the summary short."

    And:

    "Keep the summary short because the manager will read it between meetings."
    The second instruction gives Claude the reason for the length constraint.
    It explains the practical situation behind the request.


    Another important point is that context is not the same as the task. 
    Context explains the situation; 
    the task explains what Claude should do.




The key takeaway is simple: 
    do not give Claude only the task when the background affects the answer. 
    Explain the situation, 
    identify the audience, 
    state what the audience already knows, 
    and provide the reason when it helps clarify your goal.




In the next lecture, we will apply these prompting principles to business and technical tasks.

'''
````

# 6.. Writing prompts for business & technical Tasks


````python
'''

Hi there 👋,

In this lecture, we are going to understand how to write prompts for business tasks and technical tasks.


Business and technical tasks may both need :
    • clear context, 
    • a specific task, 
    • a suitable format, 
    • and relevant constraints. 
    
    

Let's start with business tasks ..
Business tasks include everyday activities such as writing emails, creating reports, preparing proposals, 
    and summarizing information.

For these tasks, think about three questions:

    • Who is the audience? ( CONTEXT)
    • Why are you creating the output? (GOAL/PURPOSE)
    • What format and tone should the output use? ( CONSTRAINTS , FORMAT )


    Imagine you are working as a Sales Executive. 
    At the end of the month, your manager asks you to share a sales performance update.

    You want Claude to prepare an email summarizing your team's performance.

    Weak prompt: ❌
    "Write an email about monthly sales."
    
    
    Strong prompt: ✅

    You are an experienced Sales Operations Assistant.
    Task: Draft an email summarizing monthly sales performance.

    Context:
        - Audience: Sales Manager
        - Monthly sales: ₹8,00,000
        - Sales target: ₹10,00,000
        - Achievement: 80% of target

    Purpose: Update the manager on monthly performance and suggest an area of focus for next month.

    Constraints:
        - Use a professional but simple tone.
        - Keep the email under 100 words.
        - Do not invent additional sales figures.

    Format: Include a subject line and a short email body.
    
    
    
    
    
    Why is this a better prompt?
        - Audience: Sales Manager.
        - Purpose: Share monthly sales performance.
        - Context: Actual sales were ₹8 lakh against a ₹10 lakh target.
        - Constraints: Under 150 words.
        - Format: Professional email.



You can apply the same approach to emails, meeting reports, proposals, and summaries. 
These are common business tasks, and a clear prompt should tell Claude who will read the result and why it is needed.







Now let's move to technical tasks.

    Technical tasks include writing formulas, data-cleaning scripts, and SQL queries..
    ie Technical Tasks are about no , data & CODE !


    For technical work, precision matters more than Tone..
    ( ie unlike Business Email , there is usually a 1-2 right answers here
    A small mistake in a formula, query, or script can affect the result.)

    So in Technical Tasks , instead of relying on a broad instruction, 
    •  provide the exact details Claude needs..
    •  Tell the exact Output u want

    For example : when asking for a data-cleaning script or a SQL query, include the relevant input details. 
    Where applicable, provide :
        •  column names, 
        •  sample rows, and 
        •  error messages. 
    These details help Claude understand the actual task rather than guessing what your data or code looks like.


    Weak prompt: ❌
    "Give me a formula to calculate sales."


    Strong prompt:  ✅
    You are an Excel expert.
    Task: Write an Excel formula to calculate total sales.

    Context:
        - Worksheet: Sales_Data
        - Column D contains Sales Amount.
        - Transaction records start from row 2 and end at row 100.

    Constraints:
        - Include all transaction records.
        - Do not include the header row.
        - Use a standard Excel formula.

    Format:
        Provide only the formula followed by one short explanation.






Why is this a better prompt?
    - Input: Customer transaction data.
    - Column details: Sales Amount is in column D.
    - Data range: D2 to D100.
    - Task: Calculate total sales.
    - Output: Excel formula with a one-line explanation.











Let's compare the emphasis..

    For a business task, pay particular attention to the audience, purpose, tone, and format..
    For a technical task, pay particular attention to the exact input, relevant column names or sample data,
                        error details, and the required output.



In the next lecture, we will look at what to do when a task is too large to handle reliably in one prompt..

'''
````

# 8.. Break down complex prompt


````python
'''

Hi there 👋,
In this lecture, we are going to understand how to break down complex prompts..


Q.. Why to break it ?
ANs : Bcz Bigger prompts which have multiple tasks gives weaks results , but why ?



    See When you give Claude a very large task with many goals and instructions, 
    then CLAUDE will have too little focus..
    ( ie many GOALS splits CLAUDE focus )
    Think of like this , Imagine a waiter working in a restaurant. 
    If ten different tables place their orders at the same time, the waiter may forget an item, mix up orders, 
    or miss an important instruction
    
    Something similar can happen when you give Claude too many different tasks in one large prompt.
    
    Summary :
    •  If prompt is big with multiple GOALS , CLAUDE may skip the parts or mixed up parts
    •  Hard to troubleshoot
        ( when 1 Giant answer comes back then it becomes difficult to troubleshoot it)
    
    

    Thats why it is useful to split a large task into smaller tasks. 
    This approach is called task decomposition.

    Task decomposition means taking one large task and dividing it into multiple smaller tasks 
    that Claude can handle one by one.



    Ques : How to Decompose a Task ?
    Ans : It consists of 4 steps :
        1 ) Goal
        2 ) SubTasks
        3 ) Order
        4 ) O/P
        
        
    Birthday Party Example : 
    
    1. Goal – What do you want to achieve?
            Before you split a Prompt , write Final Goal in 1 sentence
            eg , Organize a birthday party for your friend with around 20 guests within a budget of ₹95,000.
            
            
    2. Subtasks – What smaller tasks are involved?
            Break the overall goal into smaller activities:
            ( ie create 1 prompt for 1 sub-tasks)
            - Find a suitable birthday party venue.
            - Arrange food and drinks.
            - Order a birthday cake.
            - Purchase decorations and balloons.
            - Prepare and send invitations.
            
            
    3. Order – In what sequence should the tasks be completed?
            Now arrange these activities in a logical sequence.
            1. First, finalize the venue.
            2. Then, decide the food menu and order the cake.
            3. Next, purchase decorations and arrange the venue.
            4. Finally, send invitations with the party details.
            
            
    4. Output – What should the final result look like?
            The expected output is a complete birthday party plan containing:
            - Venue details and estimated cost.
            - Food and cake arrangements.
            - Decoration checklist.
            - Guest invitation list.
            - Overall budget breakdown.
        









Imagine you work as an HR Manager at a sales company..
Your company has hired five new sales employees, and you are responsible for their onboarding process..
Now, onboarding these employees involves several activities.

You need to:

    •  Prepare their joining documents.
    •  Create a training schedule.
    •  Assign a sales manager to each employee.
    •  Prepare welcome emails.
    •  Finally, create a complete onboarding plan.

Now, imagine giving all these instructions to Claude in one giant prompt and asking it to complete everything at once.
Claude has to handle several different objectives simultaneously.
Instead, let's break this large task into smaller steps..



    Step 1: Prepare the joining document checklist.
            First, ask Claude to create a checklist of documents required from the five new employees.


    Step 2: Create the training schedule.
            Once the document checklist is ready, ask Claude to prepare a five-day training schedule 
            covering company policies,  product knowledge, sales techniques, and CRM training.


    Step 3: Assign sales managers.
            Next, provide Claude with the list of available sales managers and ask it to distribute the five new employees 
            among them..


    Step 4: Draft welcome emails.
            Now ask Claude to prepare a welcome email for the new employees, explaining their joining date, 
            reporting time, and first-day instructions.


    Step 5: Prepare the final onboarding plan.
            Finally, ask Claude to combine the results from the previous steps into one complete onboarding plan.





'''
````

# 9.. Chaining Steps Together 

````python
'''

Q. What is Prompt Chaining ?
Ans : Subtasks in Separate prompt but they must be in Order..




               1 Big Prompt
       /        |         \           \
      /         |          \           \
     ↓          ↓           ↓           ↓
 Sub Task1  Sub-Task2    Sub-Task3    Sub-Task4

 @ Prompt1  →  @ prompt2  →  @ Prompt3  →  @ prompt4

 ┌──────────────────────────────────────────────────────────┐
 │ O/P of Prompt1 becomes I/P for next Prompt..             │
 └──────────────────────────────────────────────────────────┘
 
 This makes the process easier to follow and check. 
 Claude can focus on one part of the task at a time, and you can review the output before moving to the next step.

 

 ⇒ But how to pass?
    pass it in tags i.e. <venue_name>

• When we pass in tags, Claude knows
    which part is the earlier result -
    which part is the new instruction (prompt) -
 
• Before you pass O/P to the next,
  read it & make sure it is right..







               1 Big Prompt
       /        |         \           \
      /         |          \           \
     ↓          ↓           ↓           ↓
 Sub Task1  Sub-Task2    Sub-Task3    Sub-Task4

 @ Prompt1  →  @ p2  →  @ p3  →  @ p4

 ┌──────────────────────────────────────────────────────────┐
 │ O/P of Prompt1 becomes I/P for next Prompt..              │
 └──────────────────────────────────────────────────────────┘
    Before you pass O/P to the next, read it & make sure it is right..







2 ) The Draft → Review → Refine Chain

    This is one of the most common prompt-chaining patterns.
    It follows three steps:

    Step 1: Draft
    Ask Claude to create the first version of your content, such as an email or report.
    The goal is to produce an initial draft, not a perfect final version.

    Step 2: Review
    Ask Claude to evaluate the draft against specific criteria, such as tone, length, and accuracy.
    Claude now acts as an editor, identifying problems and areas for improvement.


    Step 3: Refine
    Ask Claude to revise the original draft using the feedback from the review.
    The result is an improved version that addresses the identified issues.


    Key rule: Let Claude review its own draft before refining it.







3 ) When to use Chaining and When Not To 
    Chaining is useful, but not necessary for every task.
    
    
    Chaining is useful , When u want to check results b//w steps
            OR
    When process must follow the same fixed order..
    


   e.g. in Organize a birthday party for your friend

      •  first you have to Find venue,
      •  then Arrange food and drinks.
      •  Order Cake 
      •  Purchase decorations and balloons ...
      
      


⇒ For simpler tasks, one prompt is enough..
        You can simply ask Claude to think step by step inside that single prompt.
        
⇒ For Bigger tasks, its good to use Chaining when you want to check results b/w steps


Modern Claude models can also plan and organize many tasks internally, so multiple prompts are not always necessary..



Key rule: Use Chain for control, not by habit.
            Use separate prompts when you need checkpoints or a fixed process. 
            Otherwise, keep everything in one clear prompt.








So remember the workflow: 
    •  start with one overall goal, 
    •  split it into smaller tasks, 
    •  put those tasks in order, 
    •  review each result, 
    •  and then combine the results for the final output.





'''
````

# 11.. Iterating prompt


````python
'''

Hi there 👋,
In this lecture, we are going to understand how to improve a prompt through iteration.

Anthropic treats prompting as a scientific loop,
not a one-shot guess.


                 Draft Prompt
                ↗             ↘
             Revise          Test &
                              Review
                ↖             ↙



If you write a first prompt & see what comes back
________________________________________________
                         ↓

Let's say you find out :
    Tone was off                       -> you mention tone  
    Fact was not available             -> you paste fact
    Format was not as expected         -> you mention format
    Length of O/P was not as expected  -> you mention length of O/P




The basic cycle is simple:

    Draft a prompt.
    Test the prompt and review the response & find gaps ( Tone , Lenght , Facts , Format are missing)
    Revise the prompt by filling gaps
    Run it again and compare the new response.





Let's understand why this is useful.

Imagine that you write a prompt asking Claude to draft an announcement. 
After reviewing the response, you notice that the tone is too formal. 
You may also realize that you forgot to explain the audience or include an important detail.

Rather than starting from scratch without understanding the problem, identify what is missing and revise the prompt.

There are three useful ways to improve a prompt.

1 ) Add context.
    If Claude does not understand the situation or audience, provide the missing background. 
    Explain who the response is for and what the request is intended to achieve.

2 ) Specify the format.
    If the content is suitable but the structure is not what you expected, tell Claude the format you want. 
    For example, specify bullet points, a table, or plain paragraphs.

3 ) Add real subject-matter details.
    If the response is still too generic, include the relevant facts and details about the actual subject. 
    Adding more words alone does not necessarily improve a prompt. 
    The important thing is to add useful information that helps Claude understand the task.

A useful rule is to change one thing at a time.

For example :
    first correct the context. 
    If the output is still not right, adjust the format. 
    If it still lacks the required detail, add the relevant subject-matter information. 

Changing one part at a time makes it easier to understand which revision affected the response.
Before revising, examine the output carefully.

Ask yourself:

    Is the tone right for the intended audience?
    Is the response missing important information?
    Did Claude produce or invent a detail that was not provided?
    Is the format the one I requested?

These checks help you identify what needs to change in the next prompt. Do not rewrite the prompt blindly.
First identify the gap, then make a targeted revision.





'''
````
