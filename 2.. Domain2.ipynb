# 2.. 3 Passes to Check CLAUDE Output

```python
'''



Hi there!
In this lecture, we are going to learn how to review Claude's output using a simple three-pass review method.

Now, you might wonder, why do we need three separate passes ?
Can't we simply read the output once and check everything ?

Well, when you try to check everything at the same time, you may overlook important details..
That's why it is better to review the same draft three times, with a different focus each time.

The three passes are :

    1. Completeness
    2. Accuracy
    3. Assumptions




Let's understand each one with a practical example.

    1 )  Completeness – Is everything there?
        In the first pass, we check whether Claude has included everything we asked for.
        And here, we start with our original prompt, not Claude's output.

        Let's say you asked Claude to prepare a weekly team performance report which includes  :
        •  five employees, 
        •  their completed tasks, 
        •  pending tasks, and 
        •  total working hours.


    Now, check four things:

    - Are all five employees included?
    - Are their completed tasks listed?
    - Are their pending tasks included?
    - Are total working hours mentioned?

    Q. Why do we start with completeness?
    Because Claude may produce a well-written report but still leave out something important.

    For example, if Claude includes only four employees instead of five, 
                 ie you need to compare the output with your original request to identify what's missing..
                 
    So, the rule is simple: **Compare the output with your original request, point by point.**
                              & Once you confirm that everything is included, move to the second pass.





    2 ) Accuracy – Is it correct?
    Now, in the second pass, we check whether the information Claude has provided is actually correct.
    Always Remember, Claude can produce a very professional-looking report and still get important details wrong.

    So, what should you check?

    You have to Focus on three things:

    I ) Key Facts
        Where u can focus on Checking important details such as :
                                                    employee names, task status, dates, and working hours..
        For example, 
            if Claude mentions that Rahul completed 8 tasks, verify that number against the original team tracker.


    II ) Calculations
         If Claude has calculated total working hours or team productivity percentages, 
         Always verify the important figures yourself..
         
         Bcz A calculation may look perfectly reasonable but still contain an error.

    III ) source information
         Focus on Compare Claude's report with the original information you provided, such as an Excel sheet or team tracker.
         For example, if the original tracker shows 42 pending tasks, make sure Claude's report also shows 42.




    One important rule here:

    **Never assume that a confident answer is a correct answer.**
    Whenever possible, trace important facts back to their original sources.






     3 ) Assumptions – What did Claude assume?

        Remember, Claude sometimes fills in details that you haven't provided.
        It may make assumptions about your requirements, business situation, or expected outcome.

        In this pass, ask yourself one simple question:
        **What is included in this report that I never actually provided?**

        For example:

        - Did Claude assume that an employee missed a deadline?
        - Did it assume that a pending task was delayed just because of poor performance?
        - Did it introduce a reason for low productivity without any supporting information?



        See, not every new detail is necessarily an assumption.
        Claude may use its existing knowledge or information obtained through a web search.


        Here is a Hack : 
        So, whenever you find an unfamiliar detail, Always check where it came from.
            If Claude provides a source, verify it. 
            If there is no clear source, treat that detail as unverified until you confirm it.


        Q. Why is this pass important?
            Because an assumption may not look like an error.
            The information may sound perfectly reasonable, but it may not reflect what actually happened.








    ### Key Takeaways

    Let's quickly summarize what we learned.

    We have three passes:

    1. **Completeness:** Is everything included? Compare the output with your original request.
    2. **Accuracy:** Is the information correct? Verify facts, recalculate important figures, and check original sources.
    3. **Assumptions:** Has Claude introduced anything you never provided? Identify and verify those details.

    Remember, Claude creates the first draft, but you are responsible for reviewing it before using or sharing it.

    In the next lecture, we are going to learn how to identify hallucinations in Claude's responses.
    So, let's move on!







'''
```

# 3.. Spotting Hallucinations : 

```python
'''


Hi there!

In the previous lecture, we learned how to review Claude's output using three passes: 
completeness, accuracy, and assumptions.


Now, in this lecture, we are going to focus on one of the most common problems in AI-generated content: 
    **Hallucinations.**
    

Q. -> So, what exactly is a hallucination?
Ans : A hallucination is when Claude generates information that sounds correct but is actually false, invented, 
      or does not match the information you provided..


Let's understand this with a simple example.

Imagine you ask Claude to prepare a summary of a business report.
Claude produces a very professional-looking summary and mentions a revenue figure of ₹50 lakh.
But when you check the original report, the actual revenue is ₹40 lakh.
That's a hallucination.

The important thing to understand is that Claude can present incorrect information in a very confident and convincing way.



Now, why do hallucinations happen?

See Claude generates responses by predicting what text is most likely to come next. 
Hence It does not automatically verify every fact against a reliable source.
So, sometimes, it may generate information that sounds reasonable but is not actually correct.



Let's look at some common warning signs.

    1 ) When CLAUDE provides Exact details without a source
        Pay special attention to very specific information, such as:

        - Exact figures and statistics.
        - Section numbers and legal references.
        - Research studies and publication years.
        - Direct quotations.

    For example, Claude may mention a specific section number from a company policy document.
    Before using that information, check whether the section actually exists and whether it supports Claude's statement.
    See Always Remember, a specific detail is not necessarily a verified detail.



    2 ) when You ask Claude about Recent information
        Be extra careful when asking Claude about recent events, product updates, or changes in company policies.

        Why?
            Because Claude's training knowledge has a cutoff date.
            It may not know about information introduced after that date unless 
                    it receives the information from you or retrieves it through an available search tool.
            So, whenever you are working with recent information, verify it using an up-to-date source.



    3 ) Narrow topics and exact lists
        Another warning sign is when you ask Claude about a very specific or less commonly documented topic.

        For example, imagine you ask Claude to list five official requirements from a particular company policy.
                     Claude gives you five points.

                     Now, don't automatically assume that all five are official requirements.
                     Check each point against the original policy, especially if some details are difficult to find.



    4 ) Look for Information that contradicts your source
        Hallucinations can also happen when Claude changes information from the documents you provide.
        For example, imagine you upload a sales file showing that a particular product generated ₹2 lakh revenue.

        But Claude mentions ₹2.5 lakh in its summary.
        The answer may look perfectly reasonable, but it does not match your original source.
        That's why you should always verify important figures and statements against the original documents.



        How to reduce hallucinations

        Before we finish, here are a few simple practices that can help reduce hallucinations.

        - Provide relevant source documents to CLAUDE whenever possible.
        - Ask Claude to use only the information available in those documents.
        - Ask it to identify information that is missing rather than guessing.
        - Request citations or references when working with external information.
        - Independently verify important facts before using them.


In the next part, we are going to understand two other important issues: inconsistencies and bias in Claude's responses.






'''
```

# 4.. Inconsistency & Bias  !

```python
'''


Hi there!

In the previous lecture, we learned :
    •  how to identify hallucinations
    •  where Claude generates information that sounds correct but may actually be false.

Now, in this lecture, we are going to look at two more problems: **Inconsistencies and Bias.**

Let's understand what these terms mean.

    - **Inconsistency** means different parts of an answer do not agree with each other.
    - **Bias** means the answer may present information in an unfair or one-sided way.



There are three things we are going to cover:

    1. Spotting inconsistencies.
    2. Spotting bias.
    3. Practical ways to catch and fix both problems.



Let's start with inconsistencies.

    1 ) Spotting Inconsistencies – Do the answers agree?
        An inconsistency occurs when two or more pieces of information contradict each other.

        Let's look at three common examples.

        I ) When CLAUDE generates different answers to the same question.**

            Imagine you ask Claude to calculate the total monthly expenses of your company.
                In the first response, Claude gives you ₹50,000.
                But when you ask the same question again, it gives you ₹55,000.

            Now, you have two different answers to the same question.
            This is a warning sign, and you need to verify the correct figure using your original data.



        II ) contradictions within the same answer.**

            Sometimes, Claude's answer may contradict itself..
            For example, imagine Claude prepares a table showing the monthly expenses of five departments.

            When you add the individual department expenses, the total comes to ₹5 lakh.
            But Claude mentions the final total as ₹5.5 lakh.

            So, always check whether the individual figures, totals, and conclusions agree with each other.




      III ) When CLAUDE O/P conflicts with your original documents or earlier instructions.**

            Let's say you provide Claude with a project report mentioning that the project deadline is 30 December.
            But in the final summary, Claude mentions 15 December.
            This is another inconsistency.

            So, whenever Claude restates information from your documents, 
            make sure the original meaning and details remain unchanged..







    2 ) Spotting Bias – Is the answer fair?

        Bias occurs when Claude presents information in an unfair way, makes assumptions about people,
        or focuses only on one particular viewpoint.

        Let's understand this with three examples.

        I ) When CLAUDE makes unfair assumptions about people or groups.**

            For example, imagine Claude is preparing a job advertisement for a sales manager.
                         It describes the ideal candidate as a young professional.
                         But your company never specified any age preference.
            So This is an unnecessary assumption and could make the job advertisement unfair to other candidates.



        II ) When CLAUDE considers stereotypes in wording and examples.**

            Bias can also appear in the examples Claude chooses or the language it uses.
            For example, imagine Claude prepares a training document where all managers are shown as men
                         and all administrative assistants are shown as women.

            Even if no individual sentence looks incorrect, the overall pattern may reinforce stereotypes.
            So, always review the examples and wording carefully, especially in documents involving people.




        III ) When CLAUDE generates 1-sided opinions.**

            Sometimes Claude presents only one side of a topic, even when there are multiple reasonable viewpoints.
            For example, imagine you ask Claude whether your company should introduce a work-from-home policy.

                Claude lists only the benefits, such as reduced travel time and improved flexibility.
                But it ignores possible challenges, such as communication difficulties and team coordination.

                As a result, you get an incomplete picture.
                So, when dealing with topics involving different viewpoints, 
                     ask Claude to present both the advantages and disadvantages.





    3 ) How to Catch and Fix Inconsistencies and Bias
        Now that we understand both problems, let's look at 3 practical habits that can help us identify them.

        I ) Re-run and compare
            You can run the same prompt again and compare the two responses.

            For example, if Claude gives you two different cost estimates, 
                         compare both answers and identify which figures need verification.

                        But remember, getting the same answer twice does not guarantee that the information is correct.
                        You still need to verify important facts against reliable sources.



        II ) ask CLAUDE for balanced viewpoints.**

            When discussing a business decision or a debatable topic, explicitly ask Claude to present multiple perspectives.
            For example:

            Ask Claude to explain the advantages, disadvantages, potential risks, and alternative viewpoints 
            before making a recommendation.

            This helps you avoid relying on a one-sided answer.




        III ) ask CLAUDE to provide clear formats and examples.**
            Give Claude a clear structure and relevant examples in your prompt.
            This helps Claude follow a consistent pattern when generating responses.

            For example, if you want a balanced business comparison, specify that 
                         every option must include advantages, disadvantages, costs, and risks.

            This makes it easier to compare the results and identify missing information..
            
            
            
            
            
            
            
            
            
            
            ### Key Takeaways

            Let's quickly summarize what we learned.

            1. **Inconsistencies:** Look for conflicting answers, incorrect totals, and differences between 
                 Claude's output and your original documents..
                 
            2. **Bias:** Watch for unfair assumptions, stereotypes, and one-sided viewpoints.
            
            3. **Practical habits:** Rerun and compare responses, ask for balanced viewpoints, use different reviewers, 
                 and provide clear formats and examples.

            Remember, reviewing Claude's output is not just about checking whether the information is correct. 
            We also need to check whether it is consistent and fair.



In the next lecture, we are going to learn how to fact-check Claude's output and verify important information 
using reliable sources.

So, let's move on!

Add practical prompts for each techniqueMake the bias guidance more nuanced










'''
```

# 5.. Fact Checking & verification techniques 

```python
'''

 


Hi there!

In the previous lectures, we learned how to identify hallucinations, inconsistencies, and bias in Claude's responses.

But identifying a potential problem is only half the job.
The next question is that : **How do we actually verify whether Claude's information is correct or not ?**

That's what we are going to learn in this lecture.

Let's start with an important point.
    See Not every claim requires the same type of verification.




So, let's understand some practical fact-checking techniques.

    Okay , so before understanding techniques , first is u have to :
    Prioritize What Needs to Be Checked
        First, focus on claims where an incorrect answer could cause real problems.
        For example: it could be 

            - Numbers, dates, names, and deadlines.
            - Laws, company policies, prices, and costs.
            - Quotes, references, and citations.

        Imagine Claude prepares a supplier comparison report.
        A small wording mistake may not matter much, 
        but an incorrect supplier price or contract deadline could lead to a bad business decision.

        So, prioritize the information that could cause the most damage if it were wrong.



    1 ) Use Source Checking

        The first verification technique is called source checking.
        It means going directly to the original source rather than relying on Claude's description of that source.

        Let's understand this with two examples.

        **Example 1: Legal references**

        Imagine Claude mentions a particular subsection of a government regulation.
        Instead of simply trusting Claude, always open the official regulation and 
        check whether that subsection actually exists and supports Claude's statement.



        **Example 2: Uploaded documents**
        Suppose you upload a company policy document and ask Claude to summarize its refund rules.
        Now, if Claude mentions a particular refund condition, 
             always search for that information in the original policy document.

        Here There are two things you need to check:

        - Does the source actually exist?
        - Does the source support the claim Claude has made?

        Always Remember, even a genuine source can be used to support an incorrect statement.
        So, always check the actual content.

        **The original source is what matters, not Claude's confidence.**




    2 ) Compare Answers in a New Chat
        Now, what if Claude provides a claim but doesn't give you a source?
        One additional technique is to ask the same question again in a separate chat and compare both answers.

        For example, :
        imagine Claude gives you a particular figure in one conversation.
        Open a separate chat, ask the same question, and compare the results.

        So If the answers are different, you have to identify that needs further checking.



        But, here is one Big important limitation..
        When Claude has memory enabled, then information from earlier conversations may influence a new chat.

        So, if you want a more independent comparison, 
        use an incognito chat or turn off memory where that option is available..
        
        
        But remember:

        **2 diff answers tell u where to look but 2 matching answers don't prove that the information is correct.**
        Seee Even if both answers agree, you still need a reliable source to verify important claims.




    3 ) Use Web Search and Citations
        For recent information or facts that you cannot verify using your own documents, web search can be useful.

        Claude can search the web when the search feature is available and 
        provide citations for the information it retrieves.

        But don't make one common mistake.
        

        Simply seeing a citation does not mean the claim has been verified.
        Open that link and check whether the original webpage actually supports Claude's statement or not

        Also, read the surrounding information because a short summary may miss important context.
        So Whenever possible, always prioritize official websites, government sources, and original documents.




    5 ) Verify Calculations Yourself

        Now, let's look at another important type of claim: calculations.
        Suppose Claude prepares a monthly sales report and calculates total revenue of ₹8 lakh.

        Instead of simply asking Claude whether the calculation is correct, 
        recalculate the important figures yourself using your spreadsheet or calculator.

        But Why?
        Because asking the same tool to verify its own calculation does not provide independent confirmation.
        So, whenever the numbers matter, check the calculation yourself.





    4 ) Test Recommendations

        Finally, let's talk about recommendations.
        Unlike a date or a financial figure, a recommendation is not simply true or false.

        You need to evaluate the reasoning behind it.

        For example, imagine Claude recommends that your company introduce a work-from-home policy.

        Now, ask Claude:

        **"Under what circumstances would this recommendation be the wrong choice?"**
        So This encourages Claude to consider potential disadvantages and situations 
        where the recommendation may not work.
        Then You can evaluate whether its reasoning makes sense for your actual business situation.

 
 
 
        ### Key Takeaways

        Let's quickly summarize what we learned.

        - **Source checking:** Always verify information against the original source.
        - **Compare answers:** Different answers reveal potential problems, but matching answers do not guarantee accuracy.
        - **Citations:** Open the links and check whether they actually support the claims.
        - **Calculations:** Recalculate important figures yourself.
        - **Recommendations:** Challenge the reasoning by asking when a recommendation might be wrong.



        And one final rule to remember:
        **If an important claim cannot be verified, don't blindly include it in your final work.**

        In the next lecture, we are going to explore practical validation techniques 
        to check Claude's output before using it.
        So, let's move on!








'''
```

# 6.. Understanding When Human Review Is Required

```python
'''



Hi there!
In the previous lecture, we learned how to fact-check Claude's output.

But is your own review always enough?
Not necessarily!



So, in this lecture, we are going to understand when human review is necessary.
There are three questions that can help you decide:

    1. Consequence – How much damage could a mistake cause?
    2. Reversibility – Can you correct the mistake later?
    3. Expertise – Do you have enough knowledge to verify the information yourself?

Let's understand each one.

    1 ) consequence – What happens if this is wrong?**

        For example:

        - An incorrect internal email may cause minor confusion.
        - A wrong price in a client quotation could create financial problems.
        - An incorrect statement in a legal document could have serious consequences.

So, focus on the potential damage, not how difficult the task looks.



     2 ) reversibility – Can you correct the mistake later?**

        An internal draft is easy to correct.
        But correcting an incorrect quotation after sending it to a client or a message
        after sending it to 2,000 customers can be much harder.

        So, the harder a mistake is to reverse, the more important a second review becomes.




    3 ) expertise – Can you verify the information yourself?**

        Imagine Claude prepares a technical integration plan for your company.
        If you don't understand the technical requirements, how will you verify whether the plan is correct?

        In this situation, involve a qualified person, such as a senior developer or technical architect.





    II ) High-Risk Areas That Need Human Review

        Some areas require additional care because mistakes can directly affect people's lives, rights, or finances.
        Let's look at a few examples.

        - **Legal:** Contracts, legal rights, and legal advice should be reviewed by a qualified legal professional.
        - **Healthcare:** Medical advice, treatments, and medicines require appropriate medical oversight.
        - **Finance:** Investment recommendations, loans, and tax-related advice need qualified financial review.
        - **Insurance:** Coverage, eligibility, and claim-related information require careful verification.
        - **Hiring and Housing:** Decisions affecting employment or access to housing need human oversight to ensure fairness.


    • legal 
    This includes contracts, legal rights, and legal advice.

    For example, imagine you use Claude to review a rental agreement.
    Claude may help you understand the clauses, but a qualified legal professional should review the document 
    before you rely on its legal interpretation.



    • healthcare.
    This includes medical advice, symptoms, treatments, and medicines.

    Claude can help explain general medical information, 
    but it should not replace a qualified healthcare professional when someone needs medical guidance.



    • financial 
    This includes investments, loans, taxation, and other important financial decisions.

    For example, if Claude recommends a particular investment, don't simply follow that recommendation.
    Have a qualified financial professional review it before making an important decision.



    • insurance.
    Insurance decisions involve coverage, eligibility, claims, and financial protection.

    An incorrect interpretation of an insurance policy could lead to serious problems when someone needs to make a claim.
    That's why important insurance-related advice should be reviewed by a qualified professional.



    • hiring and housing decisions.
    These decisions can directly affect someone's employment or access to housing.
    For example, if Claude helps prepare a candidate selection report, 
                 a human reviewer should check whether the evaluation is fair and based on relevant criteria.






    III ) How to Escalate the Work

        Let's take a simple example.

        Imagine Claude prepares a financial report for a client.
        You notice that it contains financial calculations and investment recommendations that you cannot independently verify.

        What should you do?

        1. Identify the information that needs expert verification.
        2. Send the report to a qualified financial professional.
        3. Ask them to review the calculations and recommendations.
        4. Make the necessary corrections before sharing the report.

        One final recommendation:

        Set your review rules in advance rather than waiting until a deadline.
        This helps you avoid skipping important checks when work becomes urgent.



        ### Key Takeaways

        Let's quickly summarize.

        Three questions help you decide whether human review is needed:

        1. **Consequence:** How much damage could a mistake cause?
        2. **Reversibility:** Can you correct it later?
        3. **Expertise:** Can you verify it yourself?


Remember, Claude can help you produce work faster, 
but you are still responsible for ensuring that the final output is safe and suitable to use.




'''
```

# 7.. Adapting Outputs for Your Audience

```python
'''



Hi there!

In the previous lecture, we learned how to refine and compare different versions of Claude's output and select the best one.
Now, there is one final decision to make.

    **In what format should Claude deliver the final output?**
    For example, should it give you a quick answer in the chat, create a separate document, 
                 or provide structured data for a spreadsheet?

    That's what we are going to learn in this lecture.




    1 ) Organize and Curate Before Sharing
        Before choosing an output format, first organize the content.
        Sometimes Claude includes unnecessary details, repeats the same points, or adds information that your reader doesn't need.

        So, follow three simple steps:

        •  Remove repetition -  Delete duplicate explanations and irrelevant information.
        •  Group related points-  Organize similar information under clear headings.
        •  Keep what matters-  Retain only the information your reader needs.

        Think of packing a suitcase for a short trip.
        You don't carry everything you own. 
        You keep only what you actually need.

        Follow the same approach when preparing Claude's output.





    2 ) Three Common Output Formats
        Now, let's understand the three output formats.


        • Inline Output – It is a Quick answer  that you see in the chat.**
           An inline answer appears directly inside your conversation with Claude.
           For example, imagine you ask:

          "What is the difference between revenue and profit?"
           Claude gives you a short explanation directly in the chat.

           You read the answer and move on.
           There is no need to create a separate document for such a simple question.





        •  Artifacts – Standalone content you can edit and reuse.**
           An artifact is a separate piece of content that you can work on, improve, and reuse.

           For example:

          - Business reports.
          - Documents.
          - Presentations.
          - Dashboards.
          - Interactive tools.

            Think of preparing a business report for your manager.
            Instead of keeping everything as a long chat response, you can create an artifact and refine it over multiple rounds.
            This is particularly useful when the content is substantial and you plan to edit or reuse it.




        • Structured Data – Information for spreadsheets and other tools.**
            Structured data means information arranged in a predefined format that other applications can easily process.
            eg

            - Tables.
            - CSV files.
            - JSON.



    For example, imagine Claude analyzes the costs of 12 hotel vendors.
                 If you want to work with that information in Excel, you can ask Claude to provide a table or CSV file.
                 Similarly, if another software application needs to process the information, you can request JSON.






    3 ) How to Choose the Right Format?

        Let's understand with four examples.

        •  You only want to read the answer:** Choose inline output.
        •  You want to edit and share a business report:** Choose an artifact.
        •  You want to analyze data in Excel:** Choose a table or CSV.
        •  You want to transfer data into another software application:** Choose JSON.

    Notice that the choice depends on how you intend to use the output.
    There is no need to create a separate document for a simple answer, 
    and there is no need to manually copy and format hundreds of data points when you can request structured data.





    Let's quickly summarize.

1. **Organize:** Remove repetition, group related information, and keep only what matters.
2. **Inline:** Use it for quick answers directly in the chat.
3. **Artifacts:** Use them for standalone content that you want to edit, reuse, or share.
4. **Structured Data:** Use tables, CSV, or JSON when information needs to be processed by spreadsheets or other applications.

    Remember this simple rule:
        **Choose the output format based on how you plan to use the information next.**










'''
```
